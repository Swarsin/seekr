# Stratified K-Fold Cross-Validation Lab

Imbalanced binary classification datasets distort plain k-fold splits when some folds receive far fewer positive examples than others. Stratified k-fold partitioning preserves the target class ratio inside every training and validation fold before hyperparameter tuning begins.

```python
def stratified_folds(labels, k=5):
    positives = [idx for idx, y in enumerate(labels) if y == 1]
    negatives = [idx for idx, y in enumerate(labels) if y == 0]
    folds = [[] for _ in range(k)]
    for pos_idx, item in enumerate(positives):
        folds[pos_idx % k].append(item)
    for neg_idx, item in enumerate(negatives):
        folds[neg_idx % k].append(item)
    return folds
```

## Hyperparameter Tuning and ROC-AUC Diagnostics

During grid search, evaluate each candidate hyperparameter tuple by averaging validation ROC-AUC scores across all k stratified folds. Because ROC-AUC integrates true-positive rate against false-positive rate across every probability threshold, it ranks classifier ordering quality without committing prematurely to a fixed decision cutoff.

```python
def select_best_hyperparameters(grid_scores):
    # Pick the setting with highest mean validation ROC-AUC.
    return max(grid_scores, key=lambda entry: sum(entry[1]) / len(entry[1]))
```
